<h5> <b> Problem statement 9 (Integer Doubler tool)</b></h5>
<h5.5>

Takes a list of 5 integers as input from the user, entered as a space-separated string.
Doubles each integer in the list.
Prints the doubled integers as a space-separated string.

Sample Input 1:
1 2 3 4 5

Sample Output 1:
2 4 6 8 10

```python
sample = list(map(int,input().split(' ')))
double =[i * 2 for i in sample]
print(*double)                  # 1 2 3 4 5
```

```python
double = []
for i in sample:
    double.append(i * 2)
print(*double)
```

Creates a dictionary where the keys are strings and the values are integers.
Prints the resulting dictionary.

Hint:

The question needs to be attempted using the concept of slicing.
Combining Data: zip(names, scores) is often used to combine data from different sources, such as combining names and scores in a list of students.

Sample Input 1:
a 1 b 2 c 3

Sample Output 1:
'{''a': 1, 'b': 2, 'c': 3}'

```python
sample = input().split()
str1= sample[::2]
str2 = sample[1::2]
intlist = list(map(int,str2))
print(dict(zip(str1, intlist)))             # a 1 b 2 c 3 d 4

# also we can convert string list to int list by
#intlist = [int(i) for i in str2]
#print(intlist)
```

<h5> <b> Problem statement 15 (palindrome with condition)</b> </h5>
<h5.5>
Write a Python script that checks if a given string is a valid palindrome. A palindrome is a word, phrase, or sequence that reads the same backward as forward. The script should ignore case and spaces.

Sample Input 1:
A man a plan a canal Panama

Sample Output 1:
True



```python
sample = input()
sample =sample.replace(' ','').lower()
palindrone = sample[::-1]
if sample == palindrone:
    print(True)
else:
    print(False)
```

<h5> <b> Problem statement 17 (Email substring Validator)</b></h5>
<h5.5>
You are developing an email validator. Given an email address and a substring, write a Python script to check if the given email address starts with the specified substring.

Sample Input 1:
user@example.com user

Sample Output 1:
True

```python
sample = input()
space = sample.index(" ")
email =sample[:space]
prefix= sample[space+1:]

print(email.startswith(prefix))         # sachinkhote45@gmail.com sachin
```
