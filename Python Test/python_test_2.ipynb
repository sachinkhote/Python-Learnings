<h5><b><center> Case Swap Transaformation

<h6>Problem:
Write a Python program that flips the case of every character in a string (uppercase → lowercase, lowercase → uppercase) without using loops or conditionals, and display the result.


```python
string = input("Enter your string to swap the case")
result = ''
for char in string:
    if char.isupper():
        result +=char.lower()
    elif char.islower():
        result+=char.upper()
    else:
        result +=char
print(result)
```

<h5><b><center> Symmetric difference

<h5> A△B =(A∪B) - (A∩B)

```python
setA = set(map(int,input().split()))
setB = set(map(int,input().split()))
print(setA, setB, end="\n")
diff= setA.symmetric_difference(setB)
if not diff:
    print("No symmetric difference")
else:
    print(*(sorted(diff)))
```

<h5> <b><center> Vowels replace and reverse

<h6> replaces all vowels (uppercase and lowercase) in a given string with the character 'x' and then prints the string in reverse.

```python
sample = input("Enter the string :")
vowels = "AEIOUaeiou"
replaced =''
print(sample)

for char in sample:
    if char in vowels:
        replaced += 'x'
    else:
        replaced +=char
    
print(replaced[::-1])
```

<h5><b><center> Matrix Flattening

<h6> Convert a matrix (list of lists) into a single list of integers.

```python
r = int(input("Enter number of rows you want"))
matrix= []

for _ in range(r):
    row= list(map(int, input().split(' ')))
    matrix.append(row)
print( "the Given matrix is: ", matrix, end ='\n')

flattened_list = []
for row in matrix:
    for i in row:
        flattened_list.append(i)

print("Flattened matrix is:",flattened_list)
```

```python
# also
r = int(input())
matrix = []

for _ in range(r):
    row = list(map(int, input().split()))
    matrix.append(row)

# 3. Flatten the matrix using list comprehension
flattened_list = [element for row in matrix for element in row]

print(flattened_list)
```

<h5><b><center> Key-Value Pair Setter

<h6> space separated string integer values. convert it into dictionary of key value pair

```python
input_data = list(input("Enter space separated string-integer values.").split(' '))
keys =[]
values =[]
print("space separated values are: ", input_data)

for i in range(0,len(input_data),2):
    keys.append(input_data[i])

for i in range(1,len(input_data),2):
    values.append(int(input_data[i]))

zipped = dict(zip(keys, values))
print("Data Converted to Dictionary",zipped)
```

<h5><b><center> Fibonacci sequence

```python
n = int(input())                # taking input to find fibonacci till number
a,b = 0,1                       # setting first two numbers
count = 0                       

while count<n:              
    print(a, end =' ')          # printing first number 
    a,b =b, a+b                 # making second number as first and next number as second
    count +=1
```

<h5><b><center> Absolute Differrence

<h6> finding absolute differrence between element and max element in list

```python
sample = list(map(int, input("enter space separated list").split(' ')))
max_element = max(sample)
result = []
print("Given elements are:",sample)
for i in sample:
    result.append(abs(i-max_element))
print(result)
```

<h5> <b><center> Matrix Transpose

```python
row = int(input("Enter Number of Rows"))
column = int(input("Enter Number of Columns"))                      # Taking the Matrix Dimension(row x column)
print("Rows:",row ,"Columns:",column)

matrix = []                                                         # Initializing empty list, which will store our original entred matrix        
for _ in range(row):
    matrix.append(list(map(int, input().split())))                  # takes space separated numbers and add it to Matrix, and takes as many rows as we have entered
print("Entered Matrix is:",matrix,sep='\n')

transposed =[]                                                      # initialize the empty list , to store transposed matrix
i = 0                                                               
while i < column:                                                   # runs until it porcesses all columns
    new_row = []                                                    # initiate the new list for transposed matrix rows
    j=0
    while j< row:                                                   # goes through every row
        new_row.append(matrix[j][i])                                # adds the element to transpose matrix row 
        j += 1
    transposed.append(new_row)                                      # adds tansposed row to transpose matrix 
    i += 1

for row in transposed:
    print(*(row))                                                   # * removes the brackets
```
