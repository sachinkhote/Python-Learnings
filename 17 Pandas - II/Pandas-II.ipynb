<h5><b><center> Operations on rows and columns in DataFrames

```python
                                        # Adding column to DF
df['Country']  ='USA'                   # adds new column name "Country" and assigns 'USA' value to all
df.head(3)
```

```python
                                                # Adding column based on another column
df['full_name'] = df['first_name'] + ' ' + df['last_name']              # this adds column at the end
print('Customer Data with Full Name\n', df.head(3))
```

```python
full_name = df['first_name'] + ' ' + df['last_name']
df.insert(3 ,'full_name2', full_name)                           # by this we can enter custom column at custom position
```

```python
df.head(3)
```

```python
                                                    # Adding Row to DF
df.loc['1000'] = [1001,'sachin','khote','sachin khote','sachin@deloitte.com','Male',None, None, None,'California','California',92127.0,'USA','sachin khote']
df.tail(3)
```

```python
                                                    # Deleting row or column
df.drop(index = '1000',inplace =True)               # removes last row 
df.tail(3)
```

```python
# deleting one or more columns

columns_to_drop = ['street_num', 'street_name','street_suffix']                 # removes three named columns
df.drop(columns = columns_to_drop , inplace=True)                               # inplace=True makes changes to original DF
df.head(3)
```

```python
# Renaming Row Labels

df.rename(index={0:"First"}).head(2)                # renames the first row lable to "First"
```

```python
# making dictionary to mapp old index with new

rename_dict = {0:"first", 1:"second",2:'third',4:'fourth'}
df.rename(index = rename_dict).head()
```

```python
# Renaming Column
df.rename(columns = {'first_name':'First Name', 'last_name':"Last Name"}).head(2)
```

```python
                                                    # Finding Unique Value along column
unique = df['gender'].unique()
print("Unique genders are: ", unique)
```

```python
# counting unique in gender
unique_count = df['gender'].nunique()
print("number of unique values in gender column:",unique_count)
```
