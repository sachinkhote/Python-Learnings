```python
                                        # type Conversion
df_purchases = pd.read_excel('customers.xlsx',sheet_name='purchases')
df_purchases.dtypes             # just becasue paid column has $ in front of numbers, pandas is treating it as object
```

```python
# converting the customer_num column to string coz we dont need any mathematical operation with that column
df_purchases['customer_num']= df_purchases['customer_num'].astype('str')
df_purchases.dtypes
```

```python
# converting the paid col to float type,by firstly removing $ by using replace function

df_purchases['paid'] = df_purchases['paid'].str.replace('$','').str.replace(',','').astype('float')
```

```python
print(df_purchases.dtypes)                  # now the paid column is converted succesfully
```

```python
# converting purch_date to date format
df_purchases.purch_date.head(2)                        # to check the date format of purch_datei.e., (yyyy-mm-dd hh:mm:ss)
```

```python
pd.to_datetime(df_purchases['purch_date'])          # as we purch_date was already in standard format. therfore we didn't need to define format exclusively
```

```python
# but if we have any other date format other than standard one then we have to use format parameter
try:
    pd.to_datetime(df_purchases['purch_date'], format = '%Y-%M-%d %H:%m:%S')        # in format we pass the format of date time we already have in our column
except Exception as e:
    print("there is a date which is in 03/13/2019 this format\n",e)
```

<h5><b><center> Join, Merge, Concat

```python
# Concat
df_customers = pd.read_excel('customers.xlsx',sheet_name='customers')
df_purchases = pd.read_excel('customers.xlsx',sheet_name='purchases')
df_products = pd.read_excel('customers.xlsx',sheet_name ='products')

combined = pd.concat([df_customers, df_purchases, df_products],axis = 0)
combined
```

```python
combined_row_wise = pd.concat([df_customers, df_purchases, df_products],axis = 1)       # if DF have diff column names , pandas will create new column
combined_row_wise
```

```python
# Merge (joins) - joins based on one or more keys.
merged_df = pd.merge(df_customers, df_purchases, on ='id', how ='inner')
merged_df.head(3)
```

```python
merged_right = pd.merge(df_customers ,df_purchases, on='id',how= 'right')
merged_right
```

```python
pd.merge(df_purchases, df_customers,how ='left',left_on='customer_num',right_on='id')   
```

```python
# OverLapping Column names

customers = pd.DataFrame({'id':[1,2,3],
                          'first_name':['sachin','dada','baba'], 
                          'product':['Book','pen','notebook']})
purchase = pd.DataFrame({'id':[2,3,4],
                         'product':['pen','notebook','pencil'],
                         'amount':[5,10,15]})
merged_df = pd.merge(customers , purchase , on='id', how='inner',suffixes =('_cust','_pur'))

print("Merged Dataset with Suffixes:\n",merged_df)
```

```python
# Joining

joined_df = df_customers.set_index('id').join(df_purchases)
print("Joined DataFrame\n",joined_df)
```

```python
# check for missing values in each column:

missing_in_col =df_customers.isnull().any()
print("missing values in each column:\n",missing_in_col)
```

```python
# to find number of NaN/ missing values in column,then we use sum() along with isnull()

nan_counts = df_customers.isnull().sum()
print("number of NaN values per attribute:",nan_counts)
```

```python
print(df_customers['email'].isnull().sum())             # for any particular column
```

```python
# to find NaN count for each row

nan_rowise = df_customers.isnull().sum(axis=1)          # axis 1 for horizontal and 1 for vertical
print(nan_rowise)
```

```python
# Replacing Non-convertible values with pd.NA 
df_purchases['paid'].replace('unknown',pd.NA)
```

```python
# or 
df_purchase_copy = df_purchases.copy()
df_purchase_copy['paid']=pd.to_numeric(df_purchase_copy['paid'],errors ='coerce')
```

```python
df_purchase_copy.head(3)       # all converted to NaN coz all values had $ infront of it (so those values cannot be converted to numeric)
```

```python
# 'mixed' tells Pandas to handle both 2025/09/23 and 2024-05-13 formats
# 'coerce' turns 'unknown' into NaT automatically

df_purchase_copy['purch_date'] = pd.to_datetime(df_purchase_copy['purch_date'], format='mixed', errors='coerce')
df_purchase_copy
```

```python
                                                # Dropping Missing Values
# dropping rows with missing values
customers_copy = df_customers.copy()

customers_dropped_row = customers_copy.dropna()     # by default it drops rows , if we want to drop any missing value colm we have to change axis 
customers_dropped_row                               # returns only 418 rows , which means there are lot of rows which are dropped 
```

```python
customers_dropped_colm = customers_copy.dropna(axis =1)
customers_dropped_colm.head(3)
```

```python
                                                # Imputing Missing Values
df_customers['email'].fillna('Unknown',inplace =True)
df_customers.tail(3)                            # we can clearly see the Unknown is placed at missing values place
```

```python
# filling missing values in gender column with mode(most frequent value)

gender_mode = df_customers['gender'].mode()[0]          
#mode(), returns series as 'Male' for which to acces its element we used [0] ,
# [0] to extract mode value from pandas series
df_customers['gender'].fillna(gender_mode , inplace =True)
df_customers 
```

```python
missing_street = df_customers[df_customers['street_num'].isna()]
missing_street.head()
```

```python
df_customers['street_num'].fillna(method='ffill',inplace=True)
df_customers.head()                         # see id row 2 and 5 values in street_num got filled with previous values
```

```python
df_customers['street_name'].fillna(method='bfill',inplace=True)
df_customers.head()                                     # see street_name of id row 2 got filled with its next value which is "Arkansas"
```

```python
                    # also we can write it as 
df_customers['street_suffix'] = df_customers['street_suffix'].ffill()
df_customers.head()                                     # see this way also works the same and wont raise warning aswell.
```

```python
# similarly we will fill all the remaining missign values by checking them first
df_customers.isna().sum()
```

```python
df_customers['city'] = df_customers['city'].bfill()
df_customers['state']= df_customers['state'].bfill()
df_customers['postcode'] = df_customers['postcode'].bfill()

df_customers.isna().sum()               # all 0 , good to go
```

```python
                                                # dropping Duplicates
customer_unique = df_customers.drop_duplicates(subset = 'first_name')  # removes duplicates based on 'first_name' 
customer_unique                                     # data dropped down to 932 rows from 1000
```
