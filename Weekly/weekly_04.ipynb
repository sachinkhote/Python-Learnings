```python
# calculate the average age of all employees.(without using built-in functions)
def getAverage(input_list):
        sum_age=0 
        count = 0
        for i in input_list:
                sum_age+= i 
                count+=1
        return sum_age / count
```

```python
input_list = [23, 45, 12, 67, 34]
average_age = getAverage(input_list)
print("Average age:", average_age)
```

```python
# product Reviews
import math

def analyze_ratings(ratings):
        avg = sum(ratings) / len(ratings)

        squared_diff =[(i-avg) **2 for i in ratings]
                
        variance = sum(squared_diff) / len(ratings)
        std_dev = math.sqrt(variance)
        
        highest_rating = max(ratings) 
        highest_count = ratings.count(highest_rating) 

        lowest_rating = min(ratings)
        lowest_count = ratings.count(lowest_rating)  

        return {'mean_rating': round(avg,2),'std_deviation': round(std_dev,2) ,'highest_rating':{'rating': highest_rating, 'count':highest_count},'lowest_rating':{'rating':lowest_rating, 'count':lowest_count}}
```

```python
analyze_ratings([5, 4, 3, 5, 2, 1, 4, 5, 3, 2, 4])
```

```python
# Diamond pattern using recursion

def print_diamond(n , current = 1):
    if current >n:
        return
    print(' '*(n-current),end ='')
    print("*"*(2*current-1))

    print_diamond(n, current+1)

    if current <n:
        print( ' ' * (n-current), end = '')
        print('*'*(2*current-1))
    
print_diamond(5, 1)
    
```

```python
# banking application

def withdraw_money():
    try:
        balance = float(input("Enter the current balance"))
        withdraw_amt = float(input("enter the withdrawal amount"))

        if balance <=0 or withdraw_amt <=0:
            raise ValueError("Error: Balance amount and Withdrawal amount must be greater than zero.")

        if withdraw_amt > balance:
            raise ValueError("Error: Insufficient funds.")
        
        balance -=withdraw_amt
    
    except TypeError:
        print("Error: Invalid input. Please provide numeric values.")
    except ValueError as e:
        print(e)
    else:
        print(f"Transaction successful. You withdrew ${withdraw_amt}.")
    finally:
        if balance is not None:
            print(f"Remaining balance: ${balance}.")
        
```

```python
withdraw_money()
```

```python
# Temperature conversion using Lambda function

def temp_conversion():
    celsius_to_fahrenheit = lambda c: (c *9/5) +32
    return celsius_to_fahrenheit
```

```python
result =temp_conversion()
print("25 degree Celsius in Fahrenheit is:", result(25))  # Example usage
```

```python
# Random Date Genetrator
import random
from datetime import datetime, timedelta

def generate_random_date(start_date , end_date):                    # function which will take two dates as string input
    start_date_ip = datetime.strptime(start_date ,"%Y-%m-%d")       # converting string to datetime object
    end_date_ip = datetime.strptime(end_date, "%Y-%m-%d")

    delta  = (end_date_ip - start_date_ip).days                     # calculating diff in start and end date in days
    random_days = random.randint(0, delta)                          # generating random no. of days within 0 to gap between start and end (days)
    random_date = start_date_ip + timedelta(days=random_days)       # addingn random no. of days to start date, to get random date

    return random_date.strftime("%Y-%m-%d")                         # converting datetimme object back to string 
```

```python
start_date = '2020-01-01'
end_date = '2023-12-31'
random_date = generate_random_date(start_date, end_date)
print("random date generated is:", random_date)
```

```python
random_date = generate_random_date('2025-12-01','2025-12-31')
print(random_date)                                  # this will generate diff date everytime
```

```python
# Event Management System

from datetime import datetime

def time_until_event(current_datetime, event_datetime):
    current_dt = datetime.strptime(current_datetime, "%Y-%m-%d %H:%M:%S")
    event_dt = datetime.strptime(event_datetime, "%Y-%m-%d %H:%M:%S")

    if event_dt <current_dt:
        return "Event Already Passed."
    
    time_remaining = event_dt - current_dt
    days = time_remaining.days

    hours, remainder = divmod(time_remaining.seconds, 3600)
    minutes, seconds = divmod(remainder, 60)

    return f"Time remaining; {days}days, {hours}hours, {minutes}minutes, {seconds}seconds"

try:
    event_ip = input("enter event date")
    current_ip = input("enter current date")

    result =time_until_event(current_ip, event_ip)
    print("Event Date Time is :",event_ip)
    print("Current Date Time is :", current_ip)
    print(result)
except:
    print("Error: Invalid Date/Time Format")
```
