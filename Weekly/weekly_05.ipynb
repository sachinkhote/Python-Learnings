```python
import numpy as np
ages = np.array([45, 67, 89, 23, 56, 90, 78, 92, 51, 64])

ages[(ages>45) | (ages<18)]=0
```

```python
                                        # Cheking row existance in an array
sales_data = np.array([[200, 400, 300],
                       [300, 500, 400],
                       [500, 600, 700],
                       [100, 200, 300]])

row_to_check = [300, 500, 400]

# Write your code here
if row_to_check in sales_data[:]:
    print("Row is present.")
else:
    print("Row is not present")
```

```python
                                        # Tuple to Array(change data type of array to float)
num_tuple = ('1', '2', '3', '4')

# Write your code here
list1 = []
for char in num_tuple:
    list1.append(float(char))

print(np.array((list1)))
```

```python
                                        # Replacing Negatives in array 
temperature_data = np.array([10, -5, 30, -2, 45, -12, 60])

# Replace negative values with their positive counterpart
update = []
for num in temperature_data:
    update.append(abs(num))
print(np.array((update)))
```
