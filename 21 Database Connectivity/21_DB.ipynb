```python
from sqlalchemy import text
```

```python
result =conn.execute(text("select count(*) from customers"))
for row in result:
    print(row)
```

```python
result = conn.execute(text("select gender, avg(TotalChildren) as avg_child from customers group by gender;"))
for row in result:
    print(row)
```

```python
import pandas as pd
```

```python
df_res = pd.read_sql("select gender, avg(TotalChildren) as avg_child from customers group by gender;", engine)

# engine is the connection to the database
print(df_res)           # stored in dataframe
```

```python
df_res.plot(kind ='bar', x= 'gender', y='avg_child')
```
