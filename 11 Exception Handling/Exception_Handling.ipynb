<h5> how recursion iterates

```python
def factorial(n:int):               # we set the data type for input 
    print(n)
    if n==0:
        print(f"we reached 0, now factorial will be calculated from bottom till top")
        return 1
    answer = n* factorial(n-1)
    print(f"factorial of {n} is ",answer)
    return answer
```

```python
factorial(5)
```

<h5> Normal way of findind factorial using for loop

```python
# regular functions
def fact(n):
    result = 1
    for i in range(1,n+1):
        result *= i
    return result
```

```python
fact(5)
```

<h5> Digital root (input = 987 [ 9+8+7] = 24 [2+4] = 6)

```python
def digital_root(n):
    if n<10:                # check if single digit (base case)
        return n
    
    str_num = str(n)        # convert number to string to iterate over digits
    sum =0                  # initialize sum
    
    for i in str_num:       # iterate over each digit
        sum += int(i)           # convert digit back to int and add to sum
    return digital_root(sum)    # recursive call with new sum
```

```python
digital_root(493193)
```

<h5> prime number using recursion

```python
def prime_helper(n,i):                  # helper function to check for prime, n is number to be checked, i is divisor
     if i*i >n:                         # chcek till sqrt(n)
          return True
     if n%i ==0:                        # if divisible by any number other than 1 and itself        
          return False
     return prime_helper(n,i+1)         # check for next divisor

def is_prime(n):                        # main function to check for prime           
     if n<2:
          return False
     return prime_helper(n,2)           # start checking from divisor 2
```

```python
is_prime(29)        # this function first calls is_prime which in turn calls prime_helper, then prime_helper checks for divisibility recursively
```

```python
def oneLineFunction():

    # Lambda Function to concating the string
    concat = lambda a,b : a + b

    # Lambda Function to Convert to Uppercase
    to_upper = lambda a: a.upper() 
    
    # Lambda Function to Check if a String is Palindrome
    is_palindrome = lambda a: a == a[::-1]

    # Lambda Function to Get the Length of a String
    get_length = lambda a: len(a)

    # Lambda Function to Replace Spaces with Underscores
    replace_spaces = lambda a: a.replace(" ","_")

    # Lambda Function to Get Initials from a Full Name
    get_initials = lambda a: '.'.join([x[0] for x in a.split()])+ '.'

    # Lambda Function to Reverse a String
    reversed_string = lambda a: a[::-1]

    # Don't make any changes to the return statement
    return (concat, to_upper, is_palindrome, get_length, replace_spaces, get_initials, reversed_string)
```

```python
concat, to_upper, is_palindrome, get_length, replace_spaces, get_initials, reversed_string = oneLineFunction()
                                                                    # we called oneLineFunction to get all the lambda functions
print(concat("Hello", "World"))        # HelloWorld
print(to_upper("hello"))               # HELLO
print(is_palindrome("madam"))          # True
print(get_length("Copilot"))           # 7
print(replace_spaces("Hello World"))   # Hello_World
print(get_initials("Sachin Tendulkar"))# S.T.
print(reversed_string("Python"))       # nohtyP
```

<h5> try-except-finally

```python
try:
    n= int(input("enter a number:"))
    result = 10/n
    print(result)
except ValueError:
    print("Invalid input: enter an integer")
except Exception as e:
    print("Some other exception occurred:", e)
finally:
    print("Execution completed")            # always runs whether exception occurs or not
```

```python
def divide_numbers(x,y):
    try:
        result = x/y
    except ZeroDivisionError:
        print("error: division by zero is not allowed")
    else:
        print("division successful, result is:", result)
    finally:
        print("Execution of divide_numbers completed")
```

```python
divide_numbers(10,2)
```

```python
divide_numbers(0,2)
```

```python
divide_numbers(10,0)
```
