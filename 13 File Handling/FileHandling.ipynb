```python
info = 'name : Sachin Khote, age:22, location: India'
```

```python
file = open('personal_info.txt', 'w')           # open file in write mode
file.write(info)                                # retturn number of characters written
```

```python
type(file)
```

```python
file.close()                                   # we have to close the file , so that data is saved  , otherwise data may be lost
```

```python
file = open('personal_info.txt', 'r')        # open file in read mode
data = file.read()
print(data)                                   # print the data read from file
```

```python
file = open('personal_info.txt', 'r')          
line1 = file.readline()                         # reads the first line from file
print(line1)
```

```python
line2 = file.readline()                     # when we again call it, it reads the next line
print(line2)
```

```python
print(file.readline())                  
```

```python
print(file.readline())                # since we reached the end of the file, it returns empty string

# now the cursor is at the end of file , so to read again we have to set the cursor to the beginning
```

```python
file.seek(0)                            # set the cursor to the beginning(0 position) of the file
# or we can use close() and open() again to reset the cursor to the beginning
```

```python
all_lines = file.readlines()      # reads all the lines and returns a list of lines
print(all_lines)
```

```python
file.close()
```

```python
file.close()                            # o/p : this is line 1 
                                        #        this is line 2 
```

```python
lines = ["this is line 1", "this is line 2", "this is line 3"]
```

```python
lines = [x +'\n' for x in lines]     # adding new line character at the end of each line
lines
```

```python
file = open('log.txt', 'w')
file.writelines(lines)
file.close()                    
'''
this is line 1
this is line 2
this is line 3
'''
```

```python
file = open('log.txt', 'a')
file.write('this is new line appended')             # this mode wont overwrite the previous data, it will append the new data at the end of file
```

```python
file.close()
```

```python
file = open('log.txt', 'x')                   # creates a new file, if file already exists it raises FileExistsError    
```

```python
file = open('log_2.txt','x')
file.write("this is log 2 file , opened using 'x' mode")
file.close()
```
