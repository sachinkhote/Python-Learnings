<h5><b><center> Open and read Employee.txt file

```python
try:
    with open("Employee.txt", 'r') as file:
        lines = file.readlines()                        # read line by line
except FileNotFoundError:
    print("The file 'Employee.txt' was not found.")
    exit()
```

```python
# Extract Header and Data
header =lines[0].strip().split(',')                         # get the header/columnn names
data =[line.strip().split(',') for line in lines[1:]]       # remaining data except header(1, means second line )
print(header)
```

<h5><b><center> Handle Missing Salaries

```python
salaries =[]
for row in data:
    try:
        salaries.append(float(row[4]))              # fifth column is salary, so we used row(4) to get salary
    except ValueError:
        salaries.append(None)
```

```python
salaries[:5]
```

```python
# finding Mean of salaries Excludinng None Values
valid_salaries = [s for s in salaries if s is not None]
mean_salary = sum(valid_salaries)/ len(valid_salaries)
print(mean_salary)
```

```python
# Replacing None with Mean Salary
for i in range(len(data)):
    if salaries[i] is None:                 # checks through salaries column, if None found it enter the condition
        data[i][4] = mean_salary            # data[i][4] moves to None position of salary, and replace it with mean_salary
```

```python
data[:5]                # data after replacing
```

<h5><b><center> Department Wise Employee Count

```python
department_count ={}                            # empty dictionary for storing unidue department and its count
for row in data:
    department = row[5]                         # extracting department from 6th column 
    department_count[department] = department_count.get(department,0) + 1       #counting department wise

sorted_department =sorted(department_count.items(), key = lambda x:x[1],reverse=True)
```

```python
sorted_department
```

<h5><b><center> Gender Wise Salary Analysis

```python
gender_salary = {'Male':[],'Female':[]}

for row in data:
    gender_salary[row[3]].append(float(row[4]))         # row[3] is gender, row[4] is salary

gender_avg_salary= {}                                   # initilaized empty dictionary for avg salary
for gender, salaries in gender_salary.items():          # iterating through gender_salary dict made earlier
    average=sum(salaries)/ len(salaries)                # finding avg

    gender_avg_salary[gender]= average                  # assigning gender with its avg

print(gender_avg_salary)
```
