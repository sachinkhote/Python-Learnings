<h5> Q9

```python
# Melt the Dataframe
'''
id_vars : product, region, Month
value_vars : sales , quantity
result columns should be product ,region ,month ,metric, value
'''
```

```python
df.melt(id_vars =['product', 'region','month'],
        value_vars =['sales','quantity'],           # turning from seperate column to rows
        var_name ='Metric',                         # new col name that will hold former column names(sales, quantity)
        value_name ='value')                        # new col name that will hold corresponding numeric values from sales and quantity
# converts wide format to long format
```

<h5> Q1

```python
# merge the two dataframes
df_merge = pd.merge(df, df_region, on='region', how ='left')
df_merge
```

<h5> Q2

```python
# Add column as "Achievment_percentage" ( asssuming sale is price per item )
df_merge['Achievment_percent'] = ((df_merge['sales'] * df_merge['quantity'])/df_merge['target_sales']  )*100
```

```python
df_merge
```

```python
# (assuming sales is total sales)

df_merge["Achievment_percent"] = df_merge['sales'] /df_merge['target_sales'] *100
df_merge
```

```python
# compute average achievment per region aand list the regions in descending order

df_achievement = df_merge.groupby('region')['Achievment_percent'].mean().sort_values(ascending=False)
df_achievement
```

<h5> Q3

```python
# Identify which region is performing below 10% on average
df_below_10= df_achievement[df_achievement<10]
df_below_10
```
