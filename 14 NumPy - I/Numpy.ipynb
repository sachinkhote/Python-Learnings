```python
array_3D.shape          # 3(we have three 2-D arrays), 2(we have two 1-D arrays) and 2(means 1-D array have 2 elements)
```

<h6> <center> Indexing and Slicing

```python
print(array_2D[1,0])            # will return first element of second row
print(array_2D[1][0])           # both the ways work same 
```

```python
print(array_3D[2,1,1])              # access the 2nd element from 2nd row of third row 
print(array_3D[2][1][1])
```

```python
print(array_2D[1,:])                # access the whole second row
print(array_2D[0][:])
```

```python
print(array_3D[1,:,0:2])          # access the 2nd row column of 3d array
print(array_3D[1][:][:])
```

```python
# picking 2nd row from all slices
print(array_3D[ : ,1 , :])          # 3 4  7 8  11 12
```

```python
# analyze the memory consumption
my_array = np.array([10,40,50,60,80])
print("Item Size",my_array.itemsize)                # 8 byte, returns the total memory consumption
print("Size of array",my_array.size)                # 5 elements
print("total Memory", my_array.nbytes,"bytes")      # 8*5 , 40 bytes
```

```python
# Arithmatic Operations on NumPy arrays
array1 = np.array([10, 20, 30])
array2 = np.array([2, 5, 10])

add  =array1 + array2
sub = array1 - array2
mul = array1 * array2
div = array1 / array2
print(f"Addition: {add}")
print(f"Subtraction: {sub}")
print(f"Multiplication: {mul}")
print(f"Division: {div}")
```

```python
data = np.array([[25, 180, 75], 
                 [30, 175, 68], 
                 [35, 170, 70], 
                 [40, 165, 80]])

# 1. Extract the first two rows and the first two columns
subarray = data[0:2 ,0:2]
print("Extracted 2x2 Subarray:")
print(subarray)

# 2. Extract the last row of the array
last_row = data[3]
print(f"Last Row: {last_row}")

# 3. Extract all rows but only the second column
second_column = data[:,1]
print(f"Second Column: {second_column}")
```

```python
# weather temperature filtering

temperatures = np.array([
    25, 32, 29, 35, 28, 31, 30, 27, 33, 34, 26, 29, 30, 32, 33, 
    36, 28, 31, 30, 29, 35, 27, 28, 31, 34, 30, 29, 32, 35, 28, 
    33, 26, 29, 30, 34, 32, 36, 28, 31, 30, 29, 33, 28, 30, 31, 
    32, 34, 29, 27, 35, 31, 30, 28, 33, 34, 36, 29, 31, 30, 29, 
    35, 28, 32, 30, 27, 33, 34, 28, 29, 32, 31, 30, 27, 35, 29, 
    30, 32, 34, 31, 28, 33, 30, 29, 32, 36, 28, 35, 30, 29, 31, 
    34, 32, 33, 28, 30, 29
])

# Define the threshold
threshold = 35

# Mask indexing to filter temperatures
filtered_temperatures = temperatures[temperatures>threshold]

print(f"Filtered temperatures greater than {threshold}°C:", filtered_temperatures)
```
