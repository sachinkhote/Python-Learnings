<h5 style="text-align: center;">Max. Number</h5>

```python
# finding maximum out of numbers
def max_num(a,b,c):
    if a>=b and a>=c:
        return a
    elif b>=a and b>=c:
        return b
    else:
        return c

x,y,z =10,5,15    
maximum = max_num(x,y,z)
print(f'The maximum out of {x}, {y} and {z} is:',maximum)
```

```python
# user can enter any count of numbers
def max_num(*num):
    maximum = num[0]                # set first number as maximum
    for i in num:                   
        if i>maximum:               # comparing with number,
            maximum = i             # if greater number found, we swap it with maximum
    return maximum
```

```python
max_num(10,12,15,14,67,0,12,45)         # custom length of output
```

<h5 style="text-align: center;">Prime Number</h5>

```python
# check for prime number
def is_prime(num):
    if num<=1:
        return False

    for i in range(2,num):              # also we can use (2,int(num**0.5)+1) for faster checking of factors
        if num%i==0:
            return False
    return True
```

```python
is_prime(15)
```

```python
is_prime(537)
```


<h5 style="text-align: center;"> Palindrome </h5>

```python
# determine if given numebr is a palindrome . Eg, 12321, noon, level, rotator

def is_palindrome(X):
    str_ip = str(X)                     # converting the entered number or string to string 
    if str_ip == str_ip[::-1]:          # reverses the string ,to check
        print(f"{X} is Palindrome")     
        return True
    else:
        print(f"{X} isn't a palindrome")
        return False 
```

```python
is_palindrome(12321)
```

```python
is_palindrome("sachin")
```

```python
is_palindrome("level")
```

```python
is_palindrome("Level")          # problem occured
```

<h5 style="text-align: center;">Armstrong Number</h5>


```python
# Determining the armstrong number
def is_armstrong(n):
    str_n = str(n)                  # to easily iterate over digits
    num_digits = len(str_n)         # number of digits
    
    sum_of_powers = 0
    for i in str_n:
        sum_of_powers += int(i) ** num_digits
    return sum_of_powers == n
```

```python
is_armstrong(135)
```

```python
n= 370
print(is_armstrong(n))
```

<h5 style="text-align: center;">GCD</h5>

```python
# finding GCD (greattest common divisor) of a number, using NAIVE method(check all divisors)
def GCD(x,y):
    gcd=1                                       # setting the gcd to 1
    for i in range(1,min(x,y)+1):               # range of 1 to small number out of two numbers passed
        if x%i == 0 and y%i ==0:                # if number divides it perfectly then it means it is a divisor
            gcd =i                              # now updating that number as pcd 
    return gcd
```

```python
GCD(48,18)
```

```python
# finding GCD using Euclidean Algo.( faster than NAIVE)
def gcd(x,y):
    while y!=0:                     # runs until y becomes 0
        x,y = y ,x%y                # 
    return x
```

```python
gcd(48,18)
"""
- 48 % 18 = 12
- Replace with gcd(18, 12)              # step1 

- 18 % 12 = 6
- Replace with gcd(12, 6)               # step2

- 12 % 6 = 0
- Replace with gcd(6, 0)                # step3

- b = 0 → stop
a =6

gcd(48, 18)
 → gcd(18, 12)
   → gcd(12, 6)
     → gcd(6, 0)
       → 6

```

```python
gcd(18,48)
```

```python
# recursive euclidean method

def gcd_2(x,y):
    if y==0:
        return x
    return gcd(y,x%y)
```

```python
gcd_2(48,18)        # (18,48%18 = 12)
                    # (12,18%12 = 6)
                    # (6,12%6 = 0)      ans= 6
```

<h5 style="text-align: center;">Sum and Product</h5>

```python
# finding sum and product of passed numbers

def sum_product(x,y):
    sum_num = x+y
    prod_num = x*y
    return sum_num,prod_num
```

```python
num1 = int(input("enter an number for sum and product"))
num2 = int(input("enter an number for sum and product"))
print(f"sum and product of {num1} and {num2} is:",sum_product(num1,num2))
```

```python
# for multiple numebr of inputs
def sum_product(*x):
    sum_number = sum(x)

    product = 1
    for i in x:
        product *= i
    return sum_number , product
```

```python
sum_product(1,2,3,4,5,6,7)      
```

```python
#also we can use below method to take multiple inputs

def sum_product_2(x):
    number = sum(x)

    product = 1
    for i in x:
        product *= i
    return number , product
num = list(map(int, input("enter numbers to find SUM and PRODUCT:").split()))
print(f"SUM and PRODUCT of {num} are:", sum_product_2(num))
```

<h5> Nested User Define Function

```python
def calculate_stat(x, y):
    # Nested helper function
    # This function is only visible inside calculate_stat
    def square(n):
        return n * n
    
    # Using the nested function
    return square(x) + square(y)

result = calculate_stat(3, 4)
print(f"Sum of squares: {result}") # Output: 25
```
