### Break (exits a loop)

```python
# check for username and password using break

attempt=3

while attempt>0:
    input_user = input("enter user name:")
    input_password = input("enter password:")
    if input_user == user and input_password == password:
        print("login Successful")
        break               # once right credentials entered it will break the loop
    else:
        attempt-=1
        print("wrong credentials entered")
```

```python
# shopping list using break

shopping_list = []
while True:
    user_input =input("enter your shopping list and enter 'quit' to exit")
    if user_input =="quit":
        break
    else:
        shopping_list.append(user_input)
print(shopping_list)
```

```python
# search in shopping list
search_item = input("enter item to search in list:")
for i in shopping_list:
    if i==search_item:
        print(search_item,"found")
        break                   # whenever that item is found it will not check for next item, it will immediately exit
else:
    print("no such item present in the list")
```

```python
# check for prime number
num = int(input("enter a number to check"))

for i in range(2,num):
    if num%i == 0:
        print(f"{num} is not prime ")
        break                   # when it finds out one factor it just exits , doesn't move to next (which makes it faster)
else:
    print(f"{num} is prime")
```

### continue (skips the current iteration and move to next one)

```python
num =1
while num <=10:
    if num%2 !=0:
        num+=1
        continue
    print(num)
    num+=1
```

```python
for num in range(1,11):
    if num%2==0:        
        continue
    print(num)
```

```python
# Question - return all no.s , skip multiple of 5 + skip no.s ends with 3

n= int(input('enter the number'))

for i in range(1,1+n):
    if i%5 ==0:
        continue
    elif i%10 ==3:
        continue
    else:
        print(i)                # skips 3, 5,10,13
```

```python
# Password Validation System.
# should not have 'admin', length should be >=8, ignore password if includes spaces, stop when END entered.

print("enter password (END for exit)")
list1 =[]

while True:
    password = input()
    if 'END' in password:
        break
    elif 'admin' in password:
        print(f"password contains {password}")
        continue
    elif ' ' in password:
        print("password contains space in it")
        continue
    elif len(password)<8:
        print("Passwrod length is less than 8")
        continue
    else:
        list1.append(password)
print("Valid passwords:",list1)
```
