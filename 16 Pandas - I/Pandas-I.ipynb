<h5> <center> <b> Creating a Series

```python
                                                            # from List
data = [10, 20 , 30 ,40 , 50 ,60]
s = pd.Series(data)
print(s)
```

```python
s_custom = pd.Series(data, index = ['a','b','c','d','e','f'])
print(s_custom)

s_custom2 = pd.Series(data,['a','b','c','d','e','f'])
print(s_custom2)                                                         # both of these ways works same ,either mention the index word or not , the second attribute will be taken as index
```

<h5><b><center> Mathematical Operations on Series

```python
import pandas as pd

seriesA = pd.Series([1,2,3,4,5], index = ['a','b','c','d','e'])
seriesB = pd.Series([10,20,-10,-50,100], ['z','y','a','c','e'])
```

```python
# Addition
result  = seriesA + seriesB
print("Addition of seriesA and seriesB is:\n", result)
```

```python
# we can handle these NaN values by dropping them or fillinng them with something else
result_filled = result.fillna(0)
print("Addition result after filling NaN with 0:\n",result_filled)
```

```python
result_dropped = result.dropna()
print("Addition result after dropping NaN values:\n",result_dropped)
```

```python
# Subtraction
result = seriesA - seriesB
print("Subtraction of seriesB from seriesA:\n",result)
```

```python
# MUltiplication
result = seriesA * seriesB
print("Multiplication of seriesA and seriesB:\n",result)
```

```python
# Division
result = seriesA / seriesB
print("Division of seriesA by seriesB:\n",result)
```

```python
                                                    # Fill series with factors
# find all the factors 365, print resulting series

factors =[]
for i in range(1,366):
    if 365%i ==0:
        factors.append(i)

series = pd.Series(factors)
print(series)
```

```python
                                                    # Series Cleaning and Multiplication
# check for NaN values ,  drop NaN values, perform multiplication, if multiplication result has NaN Values replace them with -1

series1 = pd.Series([10, np.nan, 30, 40, 50, np.nan, 70])
series2 = pd.Series([2, 3, np.nan, 5, 6, 7, np.nan])


series1_NaN = series1.dropna()
series2_NaN = series2.dropna()                      # dropping NaN values

series_multi = series1_NaN * series2_NaN            # Series Multiplication

result_filled = series_multi.fillna(-1)             # filling NaN values with -1
print(result_filled)
```
