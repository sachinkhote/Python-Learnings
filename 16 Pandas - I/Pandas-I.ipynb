```python
import pandas as pd
ser_1 = pd.Series(range(5))
print(ser_1)                    # returns, index   values
```

```python
ser_2 = pd.Series(range(5), ['a','b','c','d','e'])
print(ser_2)                                            # Now we gave custom index for values, with which we can access the elements later
```

```python
ser_2['b']
```

```python
ser_2[1]                    # even if we have labeled the elements with alphabets , still we can access it with normal index numbers
```

```python
ser_2['a': :2]              # starts from 'a' till last with the step size of 2
```

```python
ser_1[0:3]              #in default index STOP items is exclusive but in custom index START and STOP both are inclusive
```

<h5> <center> <b> Creating a Series

```python
                                                            # from List
data = [10, 20 , 30 ,40 , 50 ,60]
s = pd.Series(data)
print(s)
```

```python
s_custom = pd.Series(data, index = ['a','b','c','d','e','f'])
print(s_custom)

s_custom2 = pd.Series(data,['a','b','c','d','e','f'])
print(s_custom2)                                                         # both of these ways works same ,either mention the index word or not , the second attribute will be taken as index
```

```python
try:
    s= pd.Series(data , ['a','b','c','d'])
    print(s)
except Exception as e:
    print(e)                                    # length of index and values must be same
```

```python
                                                                # from Numpy Array
import numpy as np
s = np.array([10,20,30,40])
s_custom = pd.Series(s, ['w','x','y','z'])
print(s_custom)
```

```python
                                                                # from Dictionary
data_dict = {'a':10, 'b':20, 'c':30, 'd':40}
s= pd.Series(data_dict)
print(s)                                                # we do not need to explicitely create custom index 
```

<h5><b><center> Mathematical Operations on Series

```python
import pandas as pd

seriesA = pd.Series([1,2,3,4,5], index = ['a','b','c','d','e'])
seriesB = pd.Series([10,20,-10,-50,100], ['z','y','a','c','e'])
```

```python
# Addition
result  = seriesA + seriesB
print("Addition of seriesA and seriesB is:\n", result)
```

```python
# we can handle these NaN values by dropping them or fillinng them with something else
result_filled = result.fillna(0)
print("Addition result after filling NaN with 0:\n",result_filled)
```

```python
result_dropped = result.dropna()
print("Addition result after dropping NaN values:\n",result_dropped)
```

```python
# Subtraction
result = seriesA - seriesB
print("Subtraction of seriesB from seriesA:\n",result)
```

```python
# MUltiplication
result = seriesA * seriesB
print("Multiplication of seriesA and seriesB:\n",result)
```

```python
# Division
result = seriesA / seriesB
print("Division of seriesA by seriesB:\n",result)
```
