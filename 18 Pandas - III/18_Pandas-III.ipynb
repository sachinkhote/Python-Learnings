```python
import pandas as pd
df_customers = pd.read_excel(r'D:\Coding Ninjas\Python\17 Pandas - II\customers.xlsx', sheet_name ="customers")
df_purchases = pd.read_excel(r'D:\Coding Ninjas\Python\17 Pandas - II\customers.xlsx', sheet_name ="purchases")
df_products = pd.read_excel(r'D:\Coding Ninjas\Python\17 Pandas - II\customers.xlsx', sheet_name ="products")
```

<h5><b><center> Statistical Operations

```python

df_purchases['purch_date'] = pd.to_datetime(df_purchases['purch_date'])
df_purchases['paid'] = df_purchases['paid'].str.replace('$','').str.replace(',','').astype('float')
```

```python
# Mean, Median, Mode, Max, Min, Sum, Count

print("Mean of Values for each column:\n", df_purchases.mean(),'\n\n')

numeric_cols = df_purchases.select_dtypes(include=['number'])
mean_values = numeric_cols.mean()
print("Mean of numeric columns only:\n",mean_values,'\n\n')

print("mean of specific column (paid):\n", df_purchases['paid'].mean())
```

```python
print("Median for each column of dataset:\n",df_purchases.median(),'\n\n')

numeric_cols2 = df_purchases.select_dtypes(include= ['number'])
median_value = numeric_cols2.median()
print("Median Calculation for Non-Numeric column:\n",median_value, '\n\n')
```

```python
max_values = df_purchases.max(numeric_only=True)
print('Maximum Values:\n',max_values,'\n')                       # returns results in float data type, 
print("maximum values in integer data type:\n",max_values.astype(int))

# similary we can do it for Sum() , count()
```

<h5><b><center> Group By

```python
df_products['cost']= df_products['cost'].str.replace("$",'').str.replace(",",'').astype('float')
```

```python
df_products.head()
```

```python
group_by_company =df_products.groupby('company')
avg_cost = group_by_company['cost'].mean()                  # avg cost of products group by company
total_products = group_by_company['product'].count()       # calculate total no. of products per company

print("average cost of products per company: \n",avg_cost.head())
print("\n Total cost of products per company: \n",total_products.head() )
```

```python
group_by_company['cost'].aggregate(['mean','median','std','count']).sort_index(ascending=False).head()          

# also we can use aggregate with groupby and we sort it out in descending order
```

```python
# if we want to sort the table on the basis of another column, we can use following way 

group_by_company['cost'].aggregate(['mean','median','std','count']).sort_values("mean",ascending=False).head()  

# now the index is "company" and we can reset back yo default 0,1,2,3 .... by useing reset_index()
```

```python
# Group by  "purch_date" and calculate the total "paid" amount per date
total_paid_per_date = df_purchases.groupby('purch_date')['paid'].sum()
print("TOTAL PAID AMOUNT PER DATE:\n",total_paid_per_date)
```

```python
# group by "product_num" and calculate the average 'amount purchased per product

avg_amt_per_product = df_purchases.groupby("product_num")['amount'].mean()
print("Average amount per product grouped by product number:\n",avg_amt_per_product.head())
```

```python
df_products.sort_values('cost').groupby("company")['cost']
```

```python
# Reshape Data using melt function and sort this by Salary in Descending order and retrieve the first highest record and join :Department and Country 

data = {
    'Department': ['Data Analyst', 'Data Analyst', 'Data Analyst', 'Data Analyst'],
    'India': [50000, 55000, 60000, 65000],
    'US': [70000, 75000, 80000, 85000]
}

# Write your code here

df = pd.DataFrame(data)

melted = pd.melt(df, id_vars=['Department'], value_vars = ['India','US'],var_name ='Country', value_name = "Salary").sort_values(by= ['Salary'], ascending = False)

row =melted.iloc[0]
print(row['Department']+' in '+row['Country'])
```
