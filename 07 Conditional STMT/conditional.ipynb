```python
# food item per day 
food_items_per_day ={"monday": "Soya Chaap",
                     "tuesday": "Paneer",
                     "wednesday": "Bread & Milk",
                     "thursday" : "Fruits",
                     "friday": "Junk Food"}

item = input()
item = item.lower()
if item in food_items_per_day:
    print(f"Food item for {item.capitalize()} is {food_items_per_day.get(item)}.")
else:
    print(f"There is no such food item exists for {item.capitalize()}.")
```

```python
# triangle type based on sides
X =int(input())
Y =int(input())
Z =int(input())

if X+Y >Z and Y+Z>X and X+Z>Y:
    if X==Y==Z:
        print("Equilateral Triangle")
    elif X!=Y!=Z:
        print("Scalene Triangle")
    else:
        print("Isosceles Triangle")
else:
    print("Not a Triangle")
```
