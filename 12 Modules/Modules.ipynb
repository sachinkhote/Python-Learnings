```python
# Get only the current year, month, and day separately
import datetime

today = datetime.date.today()
print("Year:", today.year)
print("Month:", today.month)
print("Day:", today.day)
```

```python
# parsing string into date time object
from datetime import datetime
date_string = '2025-12-31 14:30:00'
dt_object = datetime.strptime(date_string, '%Y-%m-%d %H:%M:%S')
print(dt_object)
```

```python
# Display today’s weekday name
import datetime
today = datetime.date.today()
print("Weekday:", today.strftime("%A"))
```

```python
# Find the date 7 days from today
from datetime import datetime, timedelta            # timedelta is used to represent the difference between two dates or times
today = datetime.today()                            # get current date and time
future_date = today + timedelta(days =7)            # add 7 days to current date
print("date after 7 days:",future_date.date())      # print only date part
```

```python
# Calculate the date 90 days ago
from datetime import datetime as dt, timedelta as td 
today = dt.today()
past_date = today - td(days=90)
print("date 90 days ago:",past_date.date())
```

```python
#Calculate the number of days between two dates
from datetime import datetime

date1 = datetime(2023, 1, 1)
date2 = datetime(2025,1,1)
diff = date2 -date1
print(f"number of days between {date1.date()} and {date2.date()} is {diff.days} days")
```

```python
# Add 2 hours and 30 minutes to the current time
from datetime import datetime, timedelta
current_time = datetime.now()
new_time = current_time + timedelta(hours =2, minutes =30)
print("new time after adding 2hours and 30 minutes is:", new_time.time())
```

```python
# Check if today is a weekend
from datetime import datetime
today =datetime.today()
if today.weekday() in (5,6):
    print("Today is a weekend. COz today is -", today.strftime("%A"))       # strftime is used to format date object into readable string
else:
    print("Today is not a weekend. Coz today is -", today.strftime("%A"))
```

```python
# Print the first Monday of the current month
import datetime as dt

today = dt.date.today()                 # get today's date (30 dec 2025)
first_day = today.replace(day=1)        # this will give first day of current month, (today = 2025-12-30, then first_day = 2025-12-01.)

while first_day.weekday() !=0:          # 0 is Monday, so loop will run until first_day is Monday
    first_day += dt.timedelta(days = 1) # increment day by 1
print("first monday of current month is:", first_day)
```

<h5> Random

```python
# Generate a random float between 0.0 and 1.0
import random
random_float =random.random()       # - it always returns a floating‑point number in the range 0.0(included), 1.0(excluded)
print(random_float)
```

```python
# generate a random integer between 1 and 10 
import random 
print("random integer :", random.randint(1,10))     #it always returns random number between 1 and 10 (both included) 
```

```python
# generate a random floating point number between 1.5 and 6.5
import random
random_float = random.uniform(1.5,6.5)   # uniform is used for generating floating point number between given range
print(random_float)
```

```python
# Pick a random element from a list
import random
fruits = ['apple','banana','orange','grape','mango']
print("Random fruit from the list is:", random.choice(fruits))      # random.choice() helps to pick random element either from list, tuple etc.
```

```python
# pick two random elements from a list (with replacement )
import random 
fruits = ['apple','banana','orange','grape','mango']
random_fruits = random.choices(fruits, k=2)                     # random.choices() incudes first element again while picking second element
print("Two random fruits from the list are:", random_fruits)  
```

```python
# Pick 2 unique random elements (without replacement) from a list
import random
fruits = ['apple','banana','orange','grape','mango']
random_fruits = random.sample(fruits, k=2)                      # random.sample() ensure that both elements are unique
print("two unique random fruits from the list are:", random_fruits)
```

```python
# shuffle list of cards
import random
cards = ['king' , 'queen', 'jack', 'ace', '10', '9']
random.shuffle(cards)                       # shuffles the list in place
print("shuffled cards are:", cards)
```

```python
# Generate a reproducible random number using seed
import random
random.seed(40)                     # setting seed value to 42
print(random.random())              # The seed “locks” the random generator to a predictable sequence 
```

```python
# simulate rolling dice (1-6)
import random 
dice_roll =random.randint(1,6)
print("Dice Rolled:", dice_roll)
```

```python
# Simulate tossing a coin (Heads/Tails)
import random
coin_toss = random.choice(['Heads', 'Tails'])
print("Coin Toss Result:", coin_toss)
```

```python
# generate a random password of length 8(letter + digits)
import random 
import string

characters = string.ascii_letters + string.digits       # string.ascii_letters has both uppercase and lowercase letters, string.digits has numbers from 0-9
password = ''

for _ in range(8):
    random_char = ''.join(random.choice(characters))
    password += random_char
print("generated password:", password)
```

```python
#also
import random
import string

characters = string.ascii_letters + string.digits
password = "".join(random.choice(characters) for _ in range(8))
print("Random password:", password)
```

```python
import math
# Sigmoid Activation Function
def sigmoid(n):
    sig = 1/ (1+ math.exp(-n))
    return sig
    
# ReLU Activation Function
def relu(n):
    return max(0,n)

# Tanh Activation Function
def tanh(n):
    nr = math.exp(n) -math.exp(-n)
    dr = math.exp(n) +math.exp(-n)
    return nr/dr
```

```python
print(sigmoid(45))
print(relu(-10))
print(tanh(0))
```
