```python
class Employee:
    company = "Deloitte"
    job_type = "Permanent"
    shift = "Night"
```

```python
type(Employee)          # returns type as a class because Employee is a class 
```

```python
emp1 = Employee()    # creating an object of the class Employee
type(emp1)           # returns type as a __main__.Employee coz emp1 is an object of the class Employee and it is created in the main module.
```

```python
emp2 = Employee()    # creating another object of the class Employee
```

```python
print(id(emp1))
print(id(emp2))     # returns different id for emp1 and emp2 because they are different objects of the class Employee
```

```python
emp1.company    # returns value of company attribute for emp1 object which is "Deloitte"
```

```python
emp2.job_type = 'Contractual' # changing the value of job_type attribute for emp2 object to "Contractual"
```

```python
class Employee:
    company = "Google"      # conastant attribute

    def __init__(self, emp_id, name, salary):
        print(self)   # returns the reference of the object which is being created
        self.emp_id = emp_id
        self.name = name
        self.salary = salary 

# here self is a reference variable which refers to current object which is being created. It is used to access the attributes and methods of the class in python. It is a convention to use self as the name of the reference variable but we can use any name we want.
```

```python
emp1 = Employee(emp_id = 101, name ='Sachin', salary = 1000000)   # creating an object of the class Employee and passing values for emp_id, name and salary attributes
```

```python
emp1.name   # returns value of name attribute for emp1 object which is "Sachin"
```

```python
emp2 = Employee(102, 'Rahul', 900000)   # creating another object of the class Employee and passing values for emp_id, name and salary attributes
```
