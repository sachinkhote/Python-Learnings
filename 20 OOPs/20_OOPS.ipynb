```python
class Employee:
    company = "New Info Tech"

    # Defining the constructor
    def __init__(self, emp_id, name, salary):
        print(self)
        self.emp_id = emp_id
        self.name = name
        self.salary = salary

    def apply_increment(self, percentage):
        self.salary = self.salary * (1 + percentage / 100)
```

<h5><b><center> Inheritance

```python
class Car:
    def __init__(self, manufacturer, engine, dimension): 
        self.manufacturer = manufacturer
        self.engine = engine
        self.dimension = dimension
```

```python
class Sedan(Car):
    def __init__(self, manufacturer, engine, dimension, cruise_control):
        super().__init__(manufacturer, engine, dimension)                   # calling the constructor of the parent class Car to initialize the attributes manufacturer, engine and dimension
        self.cruise_control = cruise_control
```

```python
class SUV(Car):
    def __init__(self, manufacturer, engine, dimension, fwd):
        super().__init__(manufacturer, engine, dimension)
        self.fwd = fwd   
```

```python
city = Sedan("Honda", "Petrol", (5,2), "Radar")
```

```python
city.manufacturer
```

<h5><b><center> Polymorphism

```python
# Base Class
class Shape:
    def __init__(self):
        pass

class Circle(Shape):                        # Circle class is inheritating from Shape Class
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius **2
```

```python
class Rectangle(Shape):
    def __init__(self, Length, Breadth):
        self.Length =Length
        self.Breadth = Breadth

    def area(self):
        return self.Length * self.Breadth
```

```python
cir_1 = Circle(5)
rect_1 = Rectangle(4,6)
```

```python
print("Area of circle with radius 5 is:", cir_1.area())
print("Area of rectangle with length 4 and breadth 6 is:", rect_1.area())
```

<h5><b> Polymorphism with inheritance

```python
class Animal:                               # Base class
    def make_sound(self):                   # This is a method which is defined in the base class Animal and it is not implemented here. It is implemented in the derived classes Dog and Cat. This is an example of polymorphism where the same method make_sound() is implemented in different ways in different classes.
        pass
class Dog(Animal):  
    def make_sound(self):
        return "Bark"
class Cat(Animal):
    def make_sound(self):
        return "Meow"
    
# Using Polymorphism
def animal_sound(animal):
    print(animal.make_sound())

dog = Dog()
cat = Cat()
```

```python
animal_sound(dog)   # Output: Bark
```

```python
animal_sound(cat)
```

```python
# no __init__ needed
class Dog(Animal):
    def make_sound(self):               # __init__ method is not needed here because we are not initializing any attributes for the Dog class. We are only implementing the make_sound() method which is defined in the base class Animal.
        return "Bark"

# with per-instance data
class Dog(Animal):
    def __init__(self, name):
        self.name = name
    def make_sound(self):
        return f"{self.name} says Bark"

d = Dog("Rex")
animal_sound(d)  # prints: Rex says Bark
```
