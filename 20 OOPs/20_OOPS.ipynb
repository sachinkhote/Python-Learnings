<h5><b><center> Inheritance

```python
class Car:
    def __init__(self, manufacturer, engine, dimension): 
        self.manufacturer = manufacturer
        self.engine = engine
        self.dimension = dimension
```

```python
class Sedan(Car):
    def __init__(self, manufacturer, engine, dimension, cruise_control):
        super().__init__(manufacturer, engine, dimension)                   # calling the constructor of the parent class Car to initialize the attributes manufacturer, engine and dimension
        self.cruise_control = cruise_control
```

```python
class SUV(Car):
    def __init__(self, manufacturer, engine, dimension, fwd):
        super().__init__(manufacturer, engine, dimension)
        self.fwd = fwd   
```

```python
city = Sedan("Honda", "Petrol", (5,2), "Radar")
```

```python
city.manufacturer
```

<h5><b><center> Polymorphism

```python
# Base Class
class Shape:
    def __init__(self):
        pass

class Circle(Shape):                        # Circle class is inheritating from Shape Class
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius **2
```

```python
class Rectangle(Shape):
    def __init__(self, Length, Breadth):
        self.Length =Length
        self.Breadth = Breadth

    def area(self):
        return self.Length * self.Breadth
```

```python
cir_1 = Circle(5)
rect_1 = Rectangle(4,6)
```

```python
print("Area of circle with radius 5 is:", cir_1.area())
print("Area of rectangle with length 4 and breadth 6 is:", rect_1.area())
```

<h5><b> Polymorphism with inheritance

```python
class Animal:                               # Base class
    def make_sound(self):                   # This is a method which is defined in the base class Animal and it is not implemented here. It is implemented in the derived classes Dog and Cat. This is an example of polymorphism where the same method make_sound() is implemented in different ways in different classes.
        pass
class Dog(Animal):  
    def make_sound(self):
        return "Bark"
class Cat(Animal):
    def make_sound(self):
        return "Meow"
    
# Using Polymorphism
def animal_sound(animal):
    print(animal.make_sound())

dog = Dog()
cat = Cat()
```

```python
animal_sound(dog)   # Output: Bark
```

```python
animal_sound(cat)
```

```python
# no __init__ needed
class Dog(Animal):
    def make_sound(self):               # __init__ method is not needed here because we are not initializing any attributes for the Dog class. We are only implementing the make_sound() method which is defined in the base class Animal.
        return "Bark"

# with per-instance data
class Dog(Animal):
    def __init__(self, name):
        self.name = name
    def make_sound(self):
        return f"{self.name} says Bark"

d = Dog("Rex")
animal_sound(d)  # prints: Rex says Bark
```

<h5><b><center>  Encapsulation

```python
class BankAccount():
    def __init__(self, initial_balance):
        self.__balance = initial_balance   # __balance is a private attribute which can only be accessed within the class BankAccount and cannot be accessed from outside the class. It is a convention to use double underscore before the name of the attribute to make it private in python.

    def deposit(self ,amount):              # creates a method to deposit money 
        if amount >0:
            self.__balance += amount
        else:
            print("Deposit amount must be positive.")
    
    def withdraw(self, amount):             # creates a method to withdraw money
        if amount > self.__balance and amount >0:
            print("Insufficient funds.")
        else:
            self.__balance -= amount

    def get_balance(self):                  # creates a method to get the balance of the account
        return self.__balance 
```

```python
account = BankAccount(1000)         # creating an object of the class BankAccount and initializing the balance to 1000     
```

```python
account.deposit(500)            
print("Now the current balance is:",account.get_balance())          

# get_balance() has been created to access the private attribute __balance from outside the class BankAccount because we cannot access __balance directly from outside the class.
```

```python
account.withdraw(200)
print("Now the current balance is:", account.get_balance())
```

```python
print(account.__balance)   # this will give an error because __balance is a private attribute and cannot be accessed from outside the class BankAccount
```

```python
# getters and Setters

class person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age
    
    def get_name(self):
        return self.__name              # getter method to access the private attribute __name from outside the class person.
    
    def set_name(self, name):
        if isinstance(name, str) and name:  # isinstance() is used check if name is of type string and name is not empty.
            self.__name = name          # setter method to set the value of the private attribute __name from outside the class person.
        else:
            print("Invalid name.")

    def get_age(self):
        return self.__age               #  similar to get_name() method.
    
    def set_age(self, age):
        self.__age = age                # similar to set_name() method, but here we are not checking if age is a valid number or not ,for simplicity.
        
```

```python
person1 = person("sachin", 25)  # creating an object of the class person and initializing the name and age attributes.  

print(person1.get_name())   # accessing the name attribute using the getter method get_name().
person1.set_name("Rahul")  # changing the name attribute using the setter method set_name().
print(person1.get_name())   # accessing the name attribute again to see the change.

print(person1.get_age())    # accessing the age attribute using the getter method get_age().
person1.set_age(30)         # changing the age attribute using the setter method set_age
print(person1.get_age())    # accessing the age attribute again to see the change.
```

<h5><b><center> Abstraction

```python
from abc import ABC, abstractmethod                 # importing ABC and abstractmethod from abc module to create an abstract class and abstract method in python.

class Shape(ABC):                               # creating an abstract class Shape which is inheriting from ABC class.
    @abstractmethod
    def area(self):                            # creating an abstract method area() which is defined in the abstract class Shape and it is implemented in the derived classes Circle and Rectangle. This is an example of abstraction where we are hiding the implementation details of the area() method from the user and only providing the interface to use the method.
        pass
    
```

```python
class Circle(Shape):                        # Circle class is inheritating from Shape Class
    def area(self, radius):
        return 3.14 * radius **2
```

```python
print(Circle().area(5))   # creating an object of the Circle class and calling the area() method to calculate the area of a circle with radius 5.
```
